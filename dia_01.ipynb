## Bloque 1: Diagnóstico Práctico

```python
# Dataset
ventas = [
    {"cliente": "Ana",    "pais": "Guatemala",   "monto": 120.50},
    {"cliente": "Luis",   "pais": "México",      "monto": 80.00},
    {"cliente": "Marta",  "pais": "Guatemala",   "monto": 45.25},
    {"cliente": "Pedro",  "pais": "Colombia",    "monto": 200.00},
    {"cliente": "Sofía",  "pais": "México",      "monto": 150.75},
    {"cliente": "Diego",  "pais": "Guatemala",   "monto": 99.99},
]
```

**Ejercicio 1 (Nivel 2): ventas por país**

**Objetivo**: obtener el total vendido por cada país.\
**Resultado requerido**: un diccionario cuyas claves sean los países y cuyos valores sean la suma de los montos de ese país.\
**Restricciones**: Python puro, sin importar bibliotecas.

```python

# obtener lista de paises sin duplicados
paises = list(set([pais['pais'] for pais in ventas]))
# calcular totales de ventas por pais e imprimir totales en un diccionario
total_ventas_por_pais = {}
for pais in paises:
    total_ventas_por_pais[pais] = sum([venta['monto'] for venta in ventas if venta['pais'] == pais])

print(total_ventas_por_pais)
```

>Se analizan los montos de ventas de Guatemala, México y Colombia, de los cuales Guatemala presenta mayores ventas totales. La variación porcentual Guatemala-México y México-Colombia es de 15.16% y 15.38% respectivamente.
El dataset analizado es limitado para poder saber si la variación porcentual indicada se mantendría con un mayor número de registros de ventas y si la variación en ventas se debe a cantidades de compras o montos. Concluir con solo 6 registros no es fiable, por lo que primero se debería obtener un registro más detallado dentro un período de ventas más amplio para poder hacer una análisis más completo que permita también conocer la aportación de cada país al total de ventas.

**Ejercicio 2 (Nivel 2): función de resumen**

**Objetivo**: escribir una función llamada resumen que reciba una lista de números.\
**Resultado requerido**: debe devolver (no imprimir) un diccionario con las claves "promedio", "minimo" y "maximo". Si la lista está vacía, debe devolver None.\
**Restricciones**: sin importar bibliotecas. Pruébala con los montos de ventas y con una lista vacía.

```python
def resumen(num_lista):
    if num_lista:
        dict_resumen = {
            'promedio': sum(num_lista)/len(num_lista),
            'minimo': min(num_lista),
            'maximo': max(num_lista)
        }
        return dict_resumen
    else:
        return None

# crear lista con todos los montos de venta
montos_ventas = [venta['monto'] for venta in ventas]

resultado1 = resumen(montos_ventas)
resultado2 = resumen([])

print(f'Resultado 1: {resultado1}')
print(f'Resultado 2: {resultado2}')
```

**Ejercicio 3 (Nivel 3): ventas con impuesto**

**Objetivo**: obtener los montos con un 12% de IVA añadido, pero solo de las ventas cuyo monto original sea mayor a 100.\
**Resultado requerido**: una lista de números redondeados a 2 decimales.\
**Restricciones**: debes resolverlo con una comprensión de listas.

```python
iva = 1.12
# calculo de montos de venta con IVA para montos superiores a 100
montos_con_IVA = [round(venta['monto']*iva, 2) for venta in ventas if venta['monto'] > 100]

print(montos_con_IVA)
```
